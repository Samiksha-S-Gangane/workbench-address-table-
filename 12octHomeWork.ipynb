## Find the topper in Phisics, Chemistry, Maths subject.

```python
student = {1: [78, 97, 86], 2: [67, 89, 90], 3: [90, 90, 94]}

print(student)

physics = max(student[1][0], student[2][0], student[3][0])
print("Physics topper marks:", physics)

maths = max(student[1][1], student[2][1], student[3][1])
print("Maths topper marks:", maths)

chemistry = max(student[1][2], student[2][2], student[3][2])
print("Chemistry topper marks:", chemistry)
```
