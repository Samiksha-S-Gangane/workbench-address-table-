```python
total_student = ["Raj", "Amit", "Jay", "Payal", "Priya", "Rahul", "Pavan","Kiran", "Prachi"]
present_student = ("Raj", "Jay", "Payal","Priya", "Pavan", "Kiran")

for student in total_student:
    if student not in present_student:
        print(student)
```

```python
## List of methods

l = ["Samiksha", "Vedika", "Bhakti", "Sanika","Shruti"]
print(type(l))

#append()=> always adds the element at the end
l.append("Swarali")
print(l)

#insert()=>adds the element in specific index
l.insert(2, 23)
print(l)

#extend()=> add multiple element at the end 
l.extend([12,16])
print(l)

#pop()=> removes the element using its index and return that element
s = l.pop(2)
print(l)
print(s)

#copy() => creates a copy list
new_l = l.copy()
print(new_l)

# remove() => removes the element by its value, not by its index
l.remove(12)
print(l)

#index() => returns the index position of the element
l.index("Vedika")
print(l)

#sort()=> arrange the list in ascending order by default
l = ["Samiksha", "Vedika", "Bhakti", "Sanika","Shruti"]
l.sort()
print(l)

#reverse() => Reverse the order of the list
l.reverse()
print(l)

#count()=> counts how many times a particular value appears
print(l.count("Samiksha"))

#clear() => removes everything from the list
l.clear()
print(l)

```
