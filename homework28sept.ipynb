```python
### Arithmetic Operation
#int
a = 15
b = 12
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)

### Comparision operator
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != b)

```

```python
### Arithmetic Operator
#Float
a = 16.5
b = 9.6
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)

### Comparison Operator
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != b)

```
